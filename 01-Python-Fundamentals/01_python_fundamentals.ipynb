# **Print Function**

```python
print('hello, world')
```

```python
# input("What's your name? ") # Input is not being stored

print("hello, Sihab") # hardcoded
```

```python
# Ask user for their name
# name = input("What's your name? ")

name = 'Sihab'

# Say hello to user
print('hello,')
print(name)
```

```python
# Print 'hello, [name]' in a single line
print('hello, ' + name)
```

```python
# Print(*objects, sep=' ', end='\n')
# Print 'hello, [name]' in a better way, taking the advantages of sep and end

# print(,end)
# print('hello, ', end="???") -> hello, ???

# Overriding 'end' parameter
print('hello,', end='')
print('Shihab')
```

```python
# print(, sep)
# print('hello,', 'shihab' , sep='???') -> hello,???shihab

# overriding 'sep' parameter
print('hello,', name, sep='')
```

```python
# Print quote
# Method-1
print('hello, "frind"')

# Method-2
print("hello, \"frind\"")
```

```python
# format string/ f string
print(f'hello, {name}')
```
